```python
from matplotlib import pyplot as plt
import torch
import numpy as np
device=torch.device('mps' if torch.backends.mps.is_available() else 'cpu')
```

```python
words=open('names.txt', 'r').read().splitlines()
print(len(words))
words[:20]
```

```python
# stoi: 从'a'~'z', '<S>', '<E>'到 0~27 的映射字典
stoi={chr(ord('a')+i):i for i in range(26)}
stoi['<S>']=26
stoi['<E>']=27

dict=torch.zeros((28, 28), dtype=torch.int32)

for w in words:
    chs=['<S>']+list(w)+['<E>']
    for w1, w2 in zip(chs, chs[1:]):
        bigram=(w1, w2)
        dict[stoi[w1], stoi[w2]]+=1
```
